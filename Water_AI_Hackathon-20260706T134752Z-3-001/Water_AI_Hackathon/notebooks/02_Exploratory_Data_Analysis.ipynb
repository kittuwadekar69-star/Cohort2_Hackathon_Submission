```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("datasets/scarcity.csv")
```

```python
#Distribution of Scarcity Levels

df["Water Scarcity Level"].value_counts().plot(kind="bar")
plt.title("Distribution of Water Scarcity Levels")
plt.xlabel("Scarcity Level")
plt.ylabel("Count")
plt.show()
```

```python
#Top water consuming countries

top = df.groupby("Country")["Total Water Consumption (Billion Cubic Meters)"].mean()
top.sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("Top Countries by Average Water Consumption")
plt.ylabel("Billion Cubic Meters")
plt.show()
```

```python
#Groundwater Depletion

ground = df.groupby("Country")["Groundwater Depletion Rate (%)"].mean()
ground.sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("Countries with Highest Groundwater Depletion")
plt.show()
```

```python
#Rainfall VS Consumption

plt.scatter(
    df["Rainfall Impact (Annual Precipitation in mm)"],
    df["Total Water Consumption (Billion Cubic Meters)"]
)

plt.xlabel("Rainfall")
plt.ylabel("Water Consumption")
plt.title("Rainfall vs Water Consumption")
plt.show()
```

```python
#Sector wise water usage

sector = df[
    [
        "Agricultural Water Use (%)",
        "Industrial Water Use (%)",
        "Household Water Use (%)"
    ]
].mean()

sector.plot(kind="bar")
plt.title("Average Water Usage by Sector")
plt.ylabel("Percentage")
plt.show()
```

```python
#Trend over time

trend = df.groupby("Year")["Total Water Consumption (Billion Cubic Meters)"].mean()

trend.plot()
plt.title("Average Water Consumption Over Time")
plt.ylabel("Billion Cubic Meters")
plt.show()
```
